```python
import pandas as pd
import numpy as np

metro_csv_filepath = "data/raw/metrodata_translated.csv"

df = pd.read_csv(metro_csv_filepath, encoding="cp949")

###uuh like basic dataset commands stuff idk <//3 uncomment whatever u need and add here ig
print(df.columns.tolist())
#print(df.describe(include="all"))
```

# **DATA CLEANING**

```python
#we check number of duplicated rows
print(df.duplicated().sum())

#check combinations of repeated data
duplicate_stuff = df[
    df.duplicated(subset=["months", "metro_lines", "station_names"], keep=False) #we add all the info into "duplicate stuff"
]

print("Duplicate records of stations:")
print(
    duplicate_stuff[["months", "metro_lines", "station_names", "data_upload_date"]].head(2
                                                         )
)

passenger_columns = [
    col for col in df.columns
    if "boarding_passengers" in col or "alighting passengers" in col
]

conflicting_parts = (
    duplicate_stuff.groupby(["months", "metro_lines", "station_names"])
    [passenger_columns].nunique()
    [passenger_columns].gt(1)
    [passenger_columns].any(axis=1)
)

print("Groups but with different passenger counts: ")
conflicting_parts.sum()
```
